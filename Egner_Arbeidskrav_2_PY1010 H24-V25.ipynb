<h2>Oppgave 5: areal og omkrets av en figur</h2>

```python
'''
Omkretsen av en sirkel er 2πr
Jeg vet diameteren (a)
Jeg vet store katet (b), men ikke hypotenusen (hyp)
hyp**2 = kat**2 + kat**2 
Jeg kjenner begge katetene (a og b)

Omkretsen av figuren er b + hyp +  π*a/2
Arealet av halvsirkelen er π*(a/2)**2
Arealet av trekanten er a*b/2

Jeg vil regne ut arealet og omkretsen i en felles funksjon
'''
```

```python
import math
import pylab

def fun_A_O_figur(a,b):
    A1 = math.pi*(a/2)**2
    A2 = a*b/2
    hyp = pylab.sqrt(b**2 + a**2)
    O = b + hyp + math.pi*a/2
    return ((A1 + A2),O)
   
inn1 = 2.5 # i cm
inn2 = 5 # i cm

(totalareal,figuromkrets) = fun_A_O_figur(inn1,inn2)

print('Totalarealet er', round(totalareal,2), 'kvadratcentimeter.')
print('Figurens ytre omkrets er', round(figuromkrets,2),'cm.')
```

<h1>Oppgave 6: kode som plotter en funksjon på intervallet [-10,10]

```python
'''
en kode som plotter funksjonen f(x)=-x**2-5 for x i intervallet [-10,10]
Jeg bør bruke np.linspace(-10,10,200)
'''
```

```python
import numpy as np
import matplotlib.pyplot as plt

def fun_kurve(x):
    y = -x**2-5
    return y

inn = np.linspace(-10,10,200)

ut = fun_kurve(inn)

plt.close('all')
plt.plot(inn,ut)
plt.title('Oppgave 6: funksjon f(x)=-x**2-5')
plt.show
```
